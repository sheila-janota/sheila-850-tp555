```python
import tensorflow as tf


x = tf.Variable(1.5)
x_alterado = tf.placeholder(shape=(), dtype=tf.float32)
x_assign = tf.assign(x, x_alterado)


with tf.Session():
    x.initializer.run()
    print(x.eval())
    x_assign.eval(feed_dict={x_alterado: 5.75})
    print(x.eval())
```
