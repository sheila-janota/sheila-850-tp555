```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

**2.) Utilizando o exemplo activation_functions.ipynb como base, cite três funções de ativação diferentes das que vimos. Plote a função e sua derivada.**
<br/>
**Resposta:** As três funções de ativação que pesquisei são:  Arcosseno (Arcsinh), Leaky rectified linear unit (Leaky ReLU) e Exponential linear unit (ELU).


```python
def leaky_relu(z):
    return np.maximum(0.1*z, z)

def elu(z):
    return 0.1*(np.exp(z)-1)
    

def derivative(f, z, eps=0.000001):
    return (f(z + eps) - f(z - eps))/(2 * eps)
```

```python
z = np.linspace(-5, 5, 200)

plt.figure(figsize=(11,4))

plt.subplot(121)
plt.plot(z, np.arcsinh(z), "c-", linewidth=2, label="ArcSinh")
plt.plot(z, elu(z), "r", linewidth=2, label="ELU")
plt.plot(z, leaky_relu(z), "b-.", linewidth=2, label="Leaky ReLU")
plt.grid(True)
plt.legend(loc="upper left", fontsize=12)
plt.title("Activation functions", fontsize=14)
plt.axis([-5, 5, -1.2, 1.2])

plt.subplot(122)
plt.plot(0, 0, "ro", markersize=5)
plt.plot(0, 0, "rx", markersize=10)
plt.plot(z, derivative(np.arcsinh, z), "c-", linewidth=2, label="ArcSinh")
plt.plot(z, derivative(elu, z), "r-", linewidth=2, label="ReLU")
plt.plot(z, derivative(leaky_relu, z), "b-.", linewidth=2, label="Leaky ReLU")
plt.grid(True)
#plt.legend(loc="center right", fontsize=14)
plt.title("Derivatives", fontsize=14)
plt.axis([-5, 5, -0.2, 1.2])

plt.show()
```
