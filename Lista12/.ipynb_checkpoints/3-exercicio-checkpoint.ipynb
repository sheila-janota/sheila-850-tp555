**3. Suponha que você tenha uma rede MLP composta por uma camada de entrada com 10 neurônios, seguida por uma camada oculta com 50 neurônios e, finalmente, uma camada de saída com 3 neurônios. Todos os neurônios usam a função de ativação ReLU.**
<br/><br/>
**Sendo assim, responda:** <br/><br/>
**a.) Qual é a dimensão da matriz de entrada X ?** R: Adimensão da matriz é de mx10. <br/><br/>
**b.) Qual a dimensão do vetor de pesos, Wh , da camada oculta e de seu vetor de bias bh ?**<br/>
R: A dimensão do vetor de pesos da camada oculta Wh é 10x50. A dimensão do vetor bias bh é 50.<br/><br/>
**c.) Qual é a dimensão do vetor de pesos, Wo , da camada de saída e de seu vetor de bias bo ?**
<br/> R: A dimensão do vetor de pesos Wo da camada de saída é 50x3.A dimensão do vetor bias bo é 3. <br/><br/>
**d.) Qual é a dimensão da matriz de saída, Y , da rede?** R: A dimensão da matriz de saída Y é mx3.<br/><br/>
**e.) Escreva a equação que calcula a matriz de saída da rede, Y , como uma função de X , Wh , bh , Wo e bo.** 
<br/> R: A equação que calcula a matriz de saída da rede fica  Y = (X · Wh + bh) · Wo + bo
