**4. Quantos neurônios você precisa na camada de saída de uma rede MLP se você deseja classificar emails em spam ou ham? Qual função de ativação você deve usar na camada de saída? Se você deseja classificar a base de dados MNIST (imagens de dígitos escritos à mão), quantos neurônios você precisa na camada de saída usando qual tipo de função de ativação?**
<br/><br/>
**Reposta:** Para se classificar e-mails em spam ou ham na camada de saída de uma rede MLP é necessário apenas um neurônio. A função de ativação a ser usada na camada de saída é a função de ativação logística.
E para classificar a base de dados MNIST seria necessário 10 neurônios na camada de saída e usar a função de ativação softmax, que gera uma probabilidade por classe.
