**5. Liste todos os hiperparâmetros que você pode ajustar em uma rede MLP? Caso você perceba que a rede MLP está sobreajustando, como você pode modificar esses hiperparâmetros para tentar resolver o problema?**
<br/><br/>
**Resposta:** Em uma rede MLP pode-se ajustar os seguintes hiperparâmetros: o número de camadas ocultas, o número de neurônios em cada camada oculta e  função de ativação usada em cada camada oculta e camada de saída. <br/>
Caso haja um sobreajuste na rede MLP, o que pode ser feito para resolver o problema é tentar reduzir o número de camadas ocultas e o número de neurônios por camada oculta.
