**1. Por que a função de ativação logística foi um ingrediente chave no treinamento das primeiras redes MLPs?**
<br/><br/>
**Reposta:** 
A função de ativação logística foi um ingrediente-chave no treinamento dos primeiros MLPs, porque a sua derivada é sempre diferente de zero e bem definida em todos os pontos, permitindo que o algortimo do gradiente descendente faça progresso a cada passo.
Se a função de ativação for uma função step onde o gradiente é igual a zero, o algortimo do gradiente descendente não consegue se mover em uma superfície plana, pois não há inclinação.
