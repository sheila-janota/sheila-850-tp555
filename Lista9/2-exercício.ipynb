```python
# Import all necessary libraries.
from sklearn.datasets import make_moons
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import numpy as np
```

```python
X, y = make_moons(n_samples=9, noise=0.30, random_state=42)
C, y = make_moons(n_samples=3, noise=0.30, random_state=42) # C = Centroids
```

```python
X[0] = [1,4]
X[1] = [4,3]
X[2] = [4,5]
X[3] = [3,6]
X[4] = [6,7]
X[5] = [3,3]
X[6] = [2,5]
X[7] = [2,2]
X[8] = [2,3]

# C  = Centroids
C[0] = [5,3]
C[1] = [1,3]
C[2] = [3,4]
```

```python
km = KMeans(n_clusters=3, init=C)

km.fit(X)
```

```python
X1 = [1,4,4,3,6,3,2,2,2]
X2 = [4,3,5,6,7,3,5,2,3]

plt.scatter(X1,X2)
plt.title('Dados de treinamento')
plt.xlabel('$x_1$')
plt.ylabel('$x_2$')
plt.show()
```

```python
C1 = []
C2 = []
print('Valores ótimos para os Centroids:')
for i in range(0,3):
    C1.append(km.cluster_centers_[i][0])
    C2.append(km.cluster_centers_[i][1])
    print('C%d: (%1.2f, %1.2f)' % (i,km.cluster_centers_[i][0],km.cluster_centers_[i][1]))
```

```python
print('Para treinar o modelo foram necessárias',km.n_iter_,'iterações')
```

```python
idx0 = np.argwhere(km.labels_ == 0)
idx1 = np.argwhere(km.labels_ == 1)
idx2 = np.argwhere(km.labels_ == 2)

fig = plt.figure(figsize=(5,5))
plt.plot(X[idx0,0], X[idx0,1], '.', label='Class 0')
plt.plot(X[idx1,0], X[idx1,1], '>', label='Class 1')
plt.plot(X[idx2,0], X[idx2,1], '^', label='Class 2')
plt.plot(C1,C2, 'P', label='Centroids')
plt.xlabel('$x_1$')
plt.ylabel('$x_2$')
plt.title('Clusters a que cada dado de treinamento pertence')
plt.legend()
plt.show()
```
