**CÁLCULOS PARA A POSSIBILIDADE DE COMPRAR:**
<br/><br/>
P(Sim|X) = P(Idade|Sim) x P(Renda|Sim) x P(Estudante|Sim) x P(Classificação_Crédito|Sim) x P(Sim)
<br/><br/>
P(Sim|X) = (2/9) x (4/9) x (6/9) x (6/9) x (9/14) = **0.028218 - Probabilidade da pessoa comprar um PC.**
<br/><br/>
0.028218 / (  0.006857 + 0.028218) =  **0.8045  ~  Há 80% de chances de comprar**
<br/><br/>

**CÁLCULOS PARA A POSSIBILIDADE DE NÃO COMPRAR:**
<br/><br/>
P(Não|X) = P(Idade|Não) x P(Renda|Não) x P(Estudante|Não) x P(Classificação_Crédito|Não) x P(Não)
<br/><br/>
P(Não|X) = (3/5) x (2/5) x (1/5) x (2/5) x (5/14) = **0.006857 - Probabilidade da pessoa não comprar um PC.**
<br/><br/>
0.006857  / (  0.028218 + 0.006857 ) =  **0.1954  ~  Há 20% de chances de não comprar**
<br/><br/>

**Baseado nas probabilidades acima encontradas, a pessoa compraria Sim o PC.**
