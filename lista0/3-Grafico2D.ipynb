```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np


M = 1000
x = np.random.uniform(size=M)
w = np.random.normal(0, 1, M)
y = (1.2 + 2.3*x + 10*w)

plt.scatter(x,y) #Gráfico de dispersão

plt.xlabel('x-axis') # Adiciona label no eixo x
plt.ylabel('y-axis') # Adiciona label no eixo y

plt.title('Grafico 2D') # Aidiciona título
plt.grid(True) # Adiciona grid ao plot

# Salva plot num ficheiro png
plt.savefig('Grafico2D.png') 
```
